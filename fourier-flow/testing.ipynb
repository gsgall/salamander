```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv("1d_csv_T_A_10001.csv")
columns = [column for column in data.columns if column.startswith("T")]
columns.sort(key=lambda x: int(x.split("_")[1]))

temperatures = data[columns].to_numpy()
```

```python
plt.figure()
plt.plot(temperatures[50:].mean(axis=0))
plt.show()
```
