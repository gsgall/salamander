```python
def d_ref(m, T_ref, alpha, omega, mu_ref, mu_1_mu_inf):
  numerator = 5 * (alpha + 1) * (alpha + 2) * (m * k_B * T_ref / np.pi)
  denominator = 4 * alpha * (5 - 2 * omega) * (7 - 2 * omega) * mu_ref * (mu_1_mu_inf)

  return np.sqrt(numerator / denominator)

T_init = 273.15
omega = 0.5
alpha = 1

mu_inf_mu_1 = 1.016034
mu_1_mu_inf =  1 / mu_inf_mu_1
mu_ref = 2.117e-5
k_B = 1.380658e-23
m = 66.3e-27

numerator =  5 * (alpha + 1) * (alpha + 2) * np.sqrt(m * k_B * T_init / np.pi)
denominator =  4 * alpha * (5 - 2 * omega) * (7 - 2 * omega) * mu_ref * (mu_1_mu_inf)
d =  np.sqrt(numerator / denominator)
print(d)
```
